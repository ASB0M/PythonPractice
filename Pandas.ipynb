Series

a pandas 1-dimentional labbeled array that can hold any datatype.
think of it like a single column in a spreadsheet

```python
data = [100,102,104,200,202]
```

```python
series = pd.Series(data, index = ["a", "b", "c","d", "e"]) #Series() = series cunstructor
print(series)
```

```python
print(series.loc["a"]) # to access location by label
```

```python
series.loc["c"] = 200
print(series)
```

```python
print(series.iloc[0]) # to access location by integer
```

```python
print(series[series >=200 ])
```

```python
calories = {'Day 1': 1400, "Day 2":2100, "Day 3":1200}
```

```python
series = pd.Series(calories)
```

```python
print(series[series > 1500])
```

Dataframes

Tabular data sturcture with rows and caulmns. 2-dimentional similer to excell sheet

```python
data = {"Name":["Asad", "Saqib", "Sabih", "Ahmed"],
        "Age": [21, 41, 26, 30]}
```

```python
df = pd.DataFrame(data, index = ["emp 1", "emp 2", "emp 3", "emp 4"]) # DataFrame() -- cunstructor
```

```python
print(df)
```

```python
print(df.loc["emp 1"])
```

```python
# new caulmn

df["Job"] = ["data scientist", "finance head", "CEO", "marcom"]
```

```python
print(df)
```

```python
# new row

new_row = pd.DataFrame([{"Name": "Ali", "Age": 32, "Job": "intern"}], index = ["emp 5"]) # passing in alist of dictionaries each for a new row
```

```python
df = pd.concat([df, new_row])
print(df)
```
