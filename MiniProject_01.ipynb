# Project: Messy Online Store Sales Analyzer

You'll build a mini data pipeline for a fictional online store: generate the data, clean it, analyze it, and report insights. It uses both NumPy and pandas, and it forces you to look things up because the data is deliberately messy.

---

## Phase 1: Generate the data with NumPy

Instead of downloading a clean dataset, create your own. Use `np.random.default_rng(seed=42)` to build three tables:

- **customers** (~1,000 rows): `customer_id`, `signup_date`, `city`, `age`
- **products** (~50 rows): `product_id`, `category`, `price`, `cost`
- **orders** (~20,000 rows): `order_id`, `customer_id`, `product_id`, `quantity`, `order_date`, `discount`

**Challenge:** Make the data realistic.
- Prices should be skewed (look up `lognormal`)
- Ages should cluster around a mean (`normal`)
- Some products should sell far more than others (the `p=` parameter in `rng.choice`)

**Then make it dirty.** Randomly inject:
- ~3% missing values (`np.nan`) in `age` and `discount`
- Some negative quantities
- Duplicate order rows
- Inconsistent city spellings ("Karachi", "karachi ", "KARACHI")
- A few absurd outliers (age = 250, price = 0)

Save everything to CSV with pandas.

---

## Phase 2: Clean it with pandas

Load the CSVs and fix everything you broke:
- Find problems using `.info()`, `.describe()`, `.isna().sum()`, `.duplicated()`
- Standardize text with `.str` methods
- Decide how to handle each issue. Drop, fill with the median, or flag? Write a one-line justification for each choice in a comment.
- Convert dates with `pd.to_datetime` and set sensible dtypes

**Research prompts:**
- What's the difference between `fillna` and `interpolate`?
- When is the median better than the mean?
- What does `errors="coerce"` do?

---

## Phase 3: Analyze

Merge the three tables (`pd.merge`), then compute `revenue` and `profit` columns. Answer these questions:

1. What are the top 10 customers by total spend?
2. What is the monthly revenue trend? (`resample` or `groupby` with `dt.to_period`)
3. Which category has the best profit margin?
4. Which city has the highest average order value?
5. Build a pivot table of category × month revenue.
6. What share of customers are repeat buyers?
7. Do discounts actually increase quantity sold? Check with `.corr()` and `groupby`.
8. Segment customers into Low/Medium/High spenders using `pd.qcut`.

---

## Phase 4: NumPy deep-dive

This is where NumPy earns its place:

- **Vectorization vs loops:** Compute profit for all orders with a Python `for` loop, then with vectorized operations. Time both with `time.perf_counter()`. The speedup is the best lesson in this project.
- **Broadcasting:** Build a 2D array of revenue (products × months) and compute each product's percentage share of its month using broadcasting, with no loops.
- **`np.where` / `np.select`:** Create a "order size" label (small/medium/large) without `apply`.
- **Moving average:** Compute a 7-day rolling revenue with NumPy (`np.convolve`), then verify it matches pandas' `.rolling(7).mean()`.

---

## Phase 5: Report

- Make 4 to 5 charts with matplotlib (revenue trend, category profit, top customers, discount vs quantity).
- Write a short `insights.md` with 3 to 5 findings and what the store should do about them.

## Rules that make it harder (in a good way)

- **No `iterrows()`** unless you can justify it. Find the vectorized or `groupby` way.
- **Use method chaining** at least once (`df.assign(...).query(...).groupby(...)`).
- **Wrap your cleaning in functions** (`clean_customers(df)`, `clean_orders(df)`) so the pipeline runs end to end with one script.
- **Keep a `learned.md`:** every time you Google something, write one line about what you learned.
