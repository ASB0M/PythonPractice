```python
# Importing libraries

import numpy as np
import pandas as pd
```

```python
# Customers Data Generator.

## the logic here is to create a dictionary first and then add data randomly using for loop.

## the data will have programmed inconsistencies so the latter data cleaning process will be implemented:
##     - The Customer_ID feild will have missing values.
##     - The Signup_Date will have inconsistent date format.
##     - The City names will have spelling inconsistencies (i.e. "Karachi", "karachi", "KARACHI").
##     - Age will have missing values.

#  Dictionary initialization.
Customers_dic = {
    # "Customer_ID" : "C1001",
    # "Signup_Date" : "13/02/2025",
    # "City"        : "Karachi",
    # "Age"         : "21"
}

# Dataset Initialization.
Customers = pd.DataFrame(columns = ["Customer_ID", "Signup_Date", "City", "Age"])

# For Customer_ID:
#   - Customer id will have no missing values at this moment.
#   - The code will increment each value at the end of the loop.
id = 1000

### Start of loop

i = 0
while i<1000:

  # --------------------------------------------------------------------------------------------------------------------

  # Signup_Date generator
  #   - the correct format of date will be day/month/year.
  #   - there will be inconsistencies in the dat format:
  #     - wrong format.
  #     - inconsistent day, months, and years.

  months = ["jan", "feb", "mar", "apr", "may", "jun", "jul", "aug", "sep", "oct", "nov", "dec"]

  ## For correct date

  def Correct_Date_Gen():
    day = np.random.randint(1, 29)
    month = np.random.choice(months)
    year = np.random.randint(2019, 2025)

    date = str(day) + "/" + month + "/" + str(year)

    return date

  ## For Incorrect Date

  def Incorrect_Date_Gen():
    day = str(np.random.randint(1, 40))
    month = np.random.choice(months)
    year =  str(np.random.randint(2019, 2050))

    date1 = day + "/" + month + "/" + year
    date2 = month + "/" + day + "/" + year
    date3 = year + "/" + month + "/" + day

    date = np.random.choice([date1, date2, date3])

    return date

  date = np.random.choice([Correct_Date_Gen(), Incorrect_Date_Gen()], p = [0.8, 0.2])

  # ----------------------------------------------------------------------------------------------------------------------

  # City generator
  ## two lists of city names:
  ##  - the first one contains correct City names.
  ##  - the second one contains incorrect names.
  ##  - A random chooser selects one from both.

  correct_city_names = np.random.choice(["Karachi", "Lahore", "Hydrabad", "Multan", "Islamabad", "Peshawar"])
  incorrect_city_names = np.random.choice(["karachi", "KARACHI", "lahore", "hydrabad", "multan", "islamabad", "peshawar", ""])

  city = np.random.choice([correct_city_names, incorrect_city_names], p=[0.9, 0.1])

  #----------------------------------------------------------------------------------------------------------------------

  # For Age
  ## inconsistencies can be:
  ##  - 0 as age
  ##  - Age above avg
  ##  - missing age

  age = np.random.choice([0, np.random.randint(18, 80), np.random.randint(90, 150)], p=[0.1, 0.8, 0.1])

  # ----------------------------------------------------------------------------------------------------------------------

  cust_id = "P" + str(id)

  # updating dictionary
  Customers_dic.update({
    "Customer_ID" : [cust_id],
    "Signup_Date" : [date],
    "City"        : [city],
    "Age"         : [age]
  })

  # Appending the new row into dataset
  Customers = pd.concat([Customers, pd.DataFrame(Customers_dic)], ignore_index = True)

  # Increment in Cust_ID
  id += 1

  # Increment in loop counter
  i += 1

### End of loop


```
